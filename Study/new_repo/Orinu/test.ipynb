## RAG

In my project, I implemented a basic RAG pipeline using LangChain, ChromaDB/FAISS, Groq, and LLaMA. First, we collected the required documents or logs and loaded them using LangChain document loaders. We then cleaned and split the data into smaller chunks using a text splitter. Each chunk was converted into a vector embedding and stored in a vector database such as ChromaDB or indexed using FAISS. When a user sends a query, the query is also converted into an embedding, and the system performs similarity search to retrieve the most relevant chunks. These retrieved chunks are then added as context to the prompt and sent to the LLaMA model through Groq for inference. Finally, the LLM generates an answer based on the retrieved context rather than relying only on its pre-trained knowledge.


“I would deploy the RAG application on AWS. The FastAPI application can be containerized using Docker and deployed on ECS with Fargate, behind an Application Load Balancer. The vector database can be hosted separately, for example ChromaDB or a managed vector database depending on the production requirement. Documents can be stored in S3, and the embeddings are generated during the ingestion process and stored in the vector store. The FastAPI service receives the user query, retrieves relevant chunks from the vector database, and sends the context to the LLM through the Groq API. We can use Redis for caching if required and CloudWatch/Prometheus for monitoring.”
