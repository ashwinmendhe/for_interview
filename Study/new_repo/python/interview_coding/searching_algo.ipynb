## 1. Linear search O(n)
Linear search checks each element sequentially until the target element is found or the list ends. Its worst-case time complexity is O(n).

```python
arr = [10, 30, 20, 50, 40]

target = 50

for i in range(len(arr)):
    if arr[i] == target:
        print("found at ", i)
        break
else:
    print("not found")
```

```python
arr = [4, 8, 2, 10, 6, 15, 3]

tar = 6
for num in range(len(arr)):
    if arr[num] == tar:
        print("found ", num)
        break
else:
    print("not found")
```

## 2. Binary search O(log n)

```python
arr = [2, 4, 6, 8, 10, 15, 20]
target = 15
left = 0
right = len(arr) -1

while left <= right:
    mid = (left+right) // 2
    if arr[mid] == target:
        print("found", mid, arr[mid])
        break
    elif arr[mid] < target:
        left = mid+1
    else:
        right = mid-1
else:
    print("not found")
```
