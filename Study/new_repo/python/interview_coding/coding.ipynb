# 1. Decorator 

```python
def decorator(func):
    def wrapper(a,b):
        if a>b:
            print(a, " is greater")
        elif a<b:
            print(b , " is greater")
        else:
            print(a, b," is equal")
        return func
    return wrapper

@decorator
def valu(a,b):
    return

valu(3,2)

```

```python
import time
def decor(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print("exe time is : ", end-start)
        return result
    return wrapper

@decor
def fn():
    time.sleep(2)

fn()
```

```python
def compress(s):
    new = ''
    c = 1
    for i in range(len(s)-1):
        if s[i] == s[i+1]:
            c += 1
        else:
            new += s[i] + str(c)
            c = 1
    new += s[len(s)-1] + str(c)
    return new
print(compress("ssswwee"))
            
```

## mearging problem

```python
arr = [(2,4),(11,16),(6,8), (4,8),(5,9), (1,3)]

def merge_interval(interval):
    sort_arr = sorted(interval, key=lambda x:x[0])
    merge = [sort_arr[0]]
    for a,b in sort_arr[1:]:
        a_last, b_last = merge[-1]
        if a<=b_last:
            merge[-1] = (a_last, max(b,b_last))
        else:
            merge.append((a,b))
    return merge
intervals = [(1,3), (3,4), (5,7)]
print(merge_interval(intervals))
```

```python
## genrator using fixed size chunks

def read_in_chunck(filename, chunk_size = 1024*1024):
    with open(filename, 'rb') as f:
        while True:
            chunk = f.read(chunk_size)
            if not chunk:
                break
            yield chunk

for chunk in read_in_chunck("data.bin"):
    print(chunk)
```

```python
def long(s):
    st = 0
    max_len = 0
    start_idx = 0
    d = {}
    for i,v in enumerate(s):
        if v in d and st <=d[v]:
            st = d[v] +1
        d[v] = i

        if i - st + 1 > max_len:
            max_len = i - st +1
            start_idx = st
    longest = s[start_idx:max_len+start_idx]

    return longest, max_len 

print(long("abcdabcbb"))
```

```python
def longest_common_prefix(strs):
    if not strs:
        return ""

    # Start with the first string
    prefix = strs[0]

    for s in strs[1:]:
        # Reduce prefix until it matches the start of s
        while not s.startswith(prefix):
            prefix = prefix[:-1]
            if not prefix:
                return ""
    return prefix

a = ["flower", "flown", "flow", "floight"]
b = ["car", "train", "bike"]
print(longest_common_prefix(a))
```

## Two sum equal to target

```python
def two_sum(numbers, target):
    seen = {}
    for i, num in enumerate(numbers):
        needed = target - num
        if needed in seen:
            return [needed, num]
        seen[num] = i

numbers = [2, 7, 11, 15]
target = 9
print(two_sum(numbers, target))
```

```python
numbers = [2, 7, 11, 15]
target = 9
d = set()
for i, num in enumerate(numbers):
    needed = target - num
    if needed in d:
        print(needed, num)
        break
    else:
        d.add(num)
```
