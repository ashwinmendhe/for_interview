## Static method

```python
class Student:
    college_name = "abc college"
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    def welcome(self):
        print("welcom: ", self.name)
    @staticmethod
    def hello():
        print("hello all")

    def get_marks(self):
        return self.marks

s1 = Student("ashwin", 55)
s1.welcome()
print(s1.get_marks())

s1.hello()
```

## getter and setter

```python
## getter: A getter is used to read/access a value.
class Employee:
    def __init__(self, name, salary):
        self._salary = salary

    def get_salary(self):
        return self._salary

emp = Employee("Ashwin", 60000)
print(emp.get_salary())

# setter : A setter is used to modify/update a value.

class Employee2:
    def __init__(self, name, salary):
        self._salary = salary

    def get_salary(self):
        return self._salary
    def set_salary(self, salary):
        if salary >= 0:
            self._salary = salary
        else:
            print("salary can not be negative")

emp2 = Employee2("ashwin", -2)
print(emp2.get_salary())
emp2.set_salary(20)
print(emp2.get_salary())
```

```python
class Emp:
    def __init__(self, name, salary):
        self.name = name
        self._salary = salary

    @property
    def salary(self):
        return self._salary

    @salary.setter
    def salary(self, value):
        if value >= 0:
            self._salary = value
        else:
            print("salary can not be negative")

e1 = Emp("ash", 1)
e1.salary = -1
```

## OverRiding OverLoading 

1. Method Overriding

Overriding happens when a child class provides its own implementation of a method that already exists in the parent class.

Parent says: "I have this method."
Child says: "I will implement it differently."

```python
class Animal:
    def sound(self):
        print("animal makes sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

animal = Animal()
animal.sound()
dog = Dog()
dog.sound()

## The Dog class overrides the sound() method of Animal.  so its a runtime 
```

```python
class Animal:

    def sound(self):
        print("Animal sound")


class Dog(Animal):

    def sound(self):
        print("Bark")


class Cat(Animal):

    def sound(self):
        print("Meow")

animals = [Dog(), Cat()]
for animal in animals:
    animal.sound()
```

2. super() with Overriding

```python
class Animal:
    def sound(self):
        print("Animal makes sound")

class Dog(Animal):
    def sound(self):
        super().sound()
        print("Dog barks")
dog = Dog()
dog.sound()
```

# 3. Method Overloading

In languages like Java or C++, you can have multiple methods with the same name but different parameters:

1. add(int, int)
2. add(int, int, int)
3. add(float, float)

this is call overloading

```python
# But Python is different.

## Python does not support traditional method overloading.

## For example, this does NOT work as traditional overloading:

class Calculator:

    def add(self, a, b):
        return a + b

    def add(self, a, b, c):
        return a + b + c

# The second add() replaces the first one.

calc = Calculator()

print(calc.add(10, 20, 30))
```

```python
# 4. How do we achieve overloading-like behavior in Python?

# We commonly use default arguments.

class Calculator:

    def add(self, a, b, c=0):
        return a + b + c

# Now:

calc = Calculator()

print(calc.add(10, 20))
print(calc.add(10, 20, 30))
```

```python
5. Overriding vs Overloading

This is the important interview comparison:

	Overriding	Overloading
Where?	Parent + Child classes	Usually same class
Method name	Same	Same
Parameters	Usually same signature	Different parameters
Purpose	Change inherited behavior	Handle different inputs
Python support	✅ Yes	❌ No traditional support
Common Python approach	Inheritance	Default args / *args
```
