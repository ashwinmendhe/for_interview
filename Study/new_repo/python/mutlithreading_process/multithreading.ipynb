```python
Multiprocessing

Multiprocessing is generally useful for CPU-bound tasks. It creates separate processes, and each 
process has its own Python interpreter, memory space, and GIL. Therefore, multiple processes can 
execute Python code on different CPU cores in parallel.

I've worked with CPU-intensive operations such as image and video processing, where multiprocessing 
can be more appropriate than multithreading.

Examples:
```

```python
Asynchronous Programming

Async programming allows a task to give control back while it's waiting.

    "This operation isn't ready yet.
Let another task run."
```

```python
What is the Event Loop?

This is especially important for FastAPI interviews.

The event loop is basically the scheduler/coordinator for asynchronous tasks.

    Imagine:

async def task1():
    await network_request()

async def task2():
    await database_request()

async def task3():
    await another_request()

The event loop manages them roughly like:

Task 1 starts
    ↓
Task 1 waits for network
    ↓
Event loop switches to Task 2
    ↓
Task 2 waits for DB
    ↓
Event loop switches to Task 3
    ↓
Task 1's network response arrives
    ↓
Event loop resumes Task 1

The key idea:

Async doesn't necessarily mean multiple things are executing on multiple CPU cores 
simultaneously. It means the program can make progress on other tasks 
instead of blocking while an operation is waiting.
```
