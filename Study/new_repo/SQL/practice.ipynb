```python
intervals = [(8,14), (1,4), (3, 7), (2,6)]

def mearge_int(intervals):
  s_int = sorted(intervals, key=lambda x: x[0])
  mer = [s_int[0]]
  for a,b in intervals[1:]:
    last_a, last_b = mer[-1]
    if a <= last_b:
      mer[-1] = (last_a, max(b, last_b))
    else:
      mer.append((a,b))
  return mer
mearge_int(intervals)
```

```python
def longest(text):
    st = 0
    start_index = 0
    max_len = 0
    d = {}
    for i, v in enumerate(text):
        if v in d and d[v] > st:
            st = d[v] +1
        d[v] = i
    if i-st+1 > max_len:
        max_len = i-st+1
        start_index = st
    long = text[start_index: max_len+start_index]
    return long, max_len

longest("abcdabcabcd") 
```
