## groupby()

Read this from left to right:

df
 ↓
group rows by department
 ↓
take salary
 ↓
calculate mean


df.groupby("COLUMN")["COLUMN_TO_CALCULATE"].AGGREGATION()

```python
## What is the average salary of each department?

df.groupby("department")["salary"].mean()
```

```python
df.groupby("department")["salary"].max()
```

```python
df.groupby("department")["performance_score"].max()
```

```python
df.groupby("department")["salary"].mean()
df.groupby("department")["salary"].max()
df.groupby("department")["salary"].count()
```

```python
df.groupby("department")["salary"].agg(["min", "max", "mean"])
```

## Q. For each department, calculate average salary and average performance score.

```python
df.groupby("department").agg({
    "salary": "mean",
    "performance_score": "mean"
})
```
