## 1. Bubble sort
After one complete pass, the largest element reaches the end.

That's why it's called Bubble Sort — large values gradually "bubble" toward the end.

```python
arr = [10, 30, 70, 40, 20]

for i in range(len(arr)):
    for j in range(len(arr)-1):
        if arr[j] > arr[j+1]:
            arr[j], arr[j+1] = arr[j+1], arr[j]

print(arr)
```

```python
arr = [5, 2, 8, 1, 3]
for i in range(len(arr)):
    for j in range(len(arr)-1-i):
        if arr[j] > arr[j+1]:
            arr[j], arr[j+1] = arr[j+1], arr[j]
print(arr)

## o(n2) time complecity 
## o(1) not creating another array so sp-ace complexity is 1 
```

## 2. selection sort
Find the smallest element and put it at the beginning. Then find the next smallest and put it in the next position.

```python
arr = [5, 2, 8, 1, 3]
for i in range(len(arr)):
    min_index=i
    for j in range(i+1, len(arr)):
        if arr[j] < arr[min_index]:
            min_index = j
    arr[i], arr[min_index] = arr[min_index], arr[i]
print(arr)

# Selection Sort time complexity is O(n²) and space complexity is O(1).
    
```

```python
arr = [64, 25, 12, 22, 11]

for i in range(len(arr)):
    min_index = i
    for j in range(i+1,len(arr)):
        if arr[j] < arr[min_index]:
            min_index = j
    arr[i], arr[min_index] = arr[min_index], arr[i]
print(arr)
```
