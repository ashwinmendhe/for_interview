A shallow copy creates a new outer object, but references to nested objects are shared between the original and copied object. A deep copy recursively copies nested objects as well, so changes to nested mutable objects in the copy don't affect the original.

```python
a = [1,2,3]
b = a
b.append(4)
print(a)
print(b)
print(id(a))
print(id(b))
```

## shallow copy

```python
import copy 
a = [1,2,3]
b = copy.copy(a)
b.append(4)

## this create a new object, this ids is different 
print(id(a), id(b))
print(a, b)
```

## shallow copy with nested list
a ──→ [   •   ,   •   ]
          ↓       ↓
        [1,2]   [3,4]
          ↑       ↑
b ──→ [   •   ,   •   ]

```python
import copy 
a = [[1,2], [3,4]]
b = copy.copy(a)
# outer list is copy but inner list is still same , because nested list is shared
print(id(a) == id(b)) # false
print(id(a[0]) == id(b[0])) # True


b[0].append(100)
print(a, b)

# a[0] and b[0] refer to the same nested list
```

## deepcopy
a → [ • , • ]
      ↓   ↓
    [1,2] [3,4]


b → [ • , • ]
      ↓   ↓
    [1,2] [3,4]

```python
import copy 
a = [[1,2], [2,3]]
b = copy.deepcopy(a)

b[0].append(100)
# These inner lists are separate objects.
print(a)
print(b)
```
