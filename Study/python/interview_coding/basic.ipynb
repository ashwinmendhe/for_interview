```python
def rev(text):
    result = ""
    for i in range(len(text)-1, -1, -1):
        result += text[i]
    return result
print(rev("ashwin"))
```

## write count of consenents and vowels

```python
def count_coc_vow(txt):
    v = c = 0
    for ch in txt:
        if ch in 'aeiou':
            v += 1
        else:
            c += 1
    return v, c
v,c = count_coc_vow('ashwin')
print(f"vowels is {v} and consenants is {c}")
```

```python
def count_coc_vow(txt):
    v = c = 0
    for ch in txt.lower():
        if ch in 'aeiou':
            v +=1
        elif ch.isalpha():
            c += 1
    return v, c

v,c = count_coc_vow('ashwin 12i3u')
print(f"vowels is {v} and consenants is {c}")
```

## Palidrom

```python

def palidrom(txt):
    result = ""
    for i in range(len(txt)-1, -1, -1):
        result += txt[i]
    if txt == result:
        return True
    else:
        return False

txt = 'abccba'
print(palidrom(txt))
            
```

```python
def palidrom(txt):
    result = ""
    for i in range(len(txt)-1, -1, -1):
        result += txt[i]
    return txt == result
       
txt = 'abccbaa'
print(palidrom(txt))
```
