## write count of consenents and vowels

```python
def count_coc_vow(txt):
    v = c = 0
    for ch in txt:
        if ch in 'aeiou':
            v += 1
        else:
            c += 1
    return v, c
v,c = count_coc_vow('ashwin')
print(f"vowels is {v} and consenants is {c}")
```

```python
def count_coc_vow(txt):
    v = c = 0
    for ch in txt.lower():
        if ch in 'aeiou':
            v +=1
        elif ch.isalpha():
            c += 1
    return v, c

v,c = count_coc_vow('ashwin 12i3u')
print(f"vowels is {v} and consenants is {c}")
```

## Palidrom

```python

def palidrom(txt):
    result = ""
    for i in range(len(txt)-1, -1, -1):
        result += txt[i]
    if txt == result:
        return True
    else:
        return False

txt = 'abccba'
print(palidrom(txt))
            
```

## Print numbers from 1 to 20:

divisible by 3 → "Fizz"
divisible by 5 → "Buzz"
divisible by both 3 and 5 → "FizzBuzz"
otherwise → print the number

```python
res = {}
for n in range(1, 21):
    if n%3 == 0 and n %5 ==0:
        print(n, "FizzBuzz")
    elif n % 3 == 0:
        print(n, "Fizz")
    elif n % 5== 0:
        print(n, "Buzz")
    else:
        print(n)
```

```python
def fizzbuzz(n):
    d = {3:'fizz', 5: 'buzz'}
    
    for i in range(1, n+1):
        result = ''
        for k,v in d.items():
            if i%k==0:
                result+=v
        if not result:
            result = i
        print(result)

fizzbuzz(20)
```

## anagram

```python
s1 = "listen"
s2 = "silentt"
l1 = []
for ch in s1:
    l1.append(ch)

for ch in s2:
    if ch not in l1:
        print("no")
        break
    else:
        l1.remove(ch)
else:

    print("yes")
```

```python
s1 = "listen"
s2 = "silent"


def freq(st):
    freq1 = {}
    for ch in st:
        if ch in freq1:
            freq1[ch] += 1
        else:
            freq1[ch] = 1
    return freq1

if len(s1) != len(s2):
    print("not")
elif freq(s1) == freq(s2):
    print("yes")
else:
    print("not")
```

## extend
extend() takes the elements from the iterable and adds them one by one.

```python
a = [1,2]
a.append([3,4])
print(a)

b = [1,2]
b.extend([3,4])
print(b)
```

```python
def flat(arr):
    result = []
    for num in arr:
        if isinstance(num,list):
            result.extend(flat(num))
        else:
            result.append(num)
    return result

arr = [20, 30, [40, 50, [1, 3, 0]], [45, 5, 67]]
print(sorted(flat(arr), reverse=True))
```

```python
a = ["flower", "flown", "flow", "floght"]
def longest_common_prefix(strs):
    if not strs:
        return ""
    prefix = strs[0]
    for s in strs[1:]:
        while not s.startswith(prefix):
            prefix = prefix[:-1]
    return prefix
print(longest_common_prefix(a))



```
