# Lambda
A lambda is a small anonymous function defined using the lambda keyword. It can take multiple arguments but contains a single expression. It is commonly used for short operations with functions such as map(), filter(), reduce(), and sorted().

```python
def add(a,b):
    return a+b

lambda a,b : a+b

# lambda arguments: expression
```

```python
sq = lambda x : x*x
print(sq(5))

add = lambda a, b : a+b
print(add(5,4))
```

## lambda with condition

```python
check = lambda x: "Even" if x%2==0 else "Odd"
print(check(10))
print(check(9))
```

## map

```python
nums = [1,2,3,4]
def sq(x):
    return x*x
result = map(sq, nums)
print( list(result))

result = map(lambda x : x*x , nums)
print(list(result))
```

## Filter

```python
def is_even(x):
    return x%2 == 0
nums = [1,2,3,4,5,6]
result = filter(is_even, nums)
print(list(result))
```

```python
nums = [1,2,3,4,5,6]
result = filter(lambda x: x%2==0 , nums)
print(list(result))
```

## reduce

```python
from functools import reduce
nums = [1,2,3,4,5]
result = reduce(lambda x, y : x+y, nums)
print(result)
```
