## Regular expression
OCR ODM 

```python
import re
text = '''Elon musk phone is 7020954476 call him if you have any tesla's reveenu 40 billian CFO no (999)-(3334)-777'''
pattern = r"\(\d{3}\)-\(\d{4}\)-\d{3}|\d{10}"
# pattern = r'\(\d{3}\)-\(\d{4}\)-\d[0-9]+'

matchs = re.findall(pattern, text)
matchs
```
