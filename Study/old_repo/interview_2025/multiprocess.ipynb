```python
import multiprocessing as mp
from multiprocessing import Process

def process_fu():
    print("this is new process")
    print("process name is : ", mp.current_process())


p1 = Process(target=process_fu)
p2 = Process(target=process_fu)

p1.start()
p2.start()
p1.join()
p2.join()
```

## threding

```python
import threading
from threading import Thread
from time import sleep

def tf():
    print("start\n")
    sleep(1)
    print("end")


th1 = Thread(target=tf)
th2 = Thread(target=tf)

th1.start()
th2.start()
threading.active_count()

```

```python
import threading
from threading import Thread
import time

def first():
    print("first occupy the room")
    time.sleep(5)
    print("first person release the room")

def second():
    print("second occupy the room")
    time.sleep(5)
    print("second person release the room")
    
```

```python
threading.active_count()
```

```python
thrd_run1 = Thread(target=first)
thrd_run2 = Thread(target=second)
thrd_run1.start()
thrd_run2.start()
thrd_run1.join()
thrd_run2.join()
threading.active_count()
```
