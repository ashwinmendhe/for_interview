```python
## for creating and switching to a new branch named 'develop'
git branch develop


git checkout develop

or 
git switch develop

## create and switch in one command
git checkout -b develop
or 
git switch -c develop

## git mearge : if we create any branch from develop branch lest say feature branch 
## so feature is the source branch and develop is the target branch
git checkout develop
git merge feature


## PR git pull notification via teams 
git pull origin develop


## git fork : to create a copy of repository in your github account

## git tag : to create a tag for a specific commit
git tag -a v1.0 -m "version 1.0"

## git revert commit_id : to revert a specific commit
git revert commit_id


## git remote
git remote -v   
git remote add origin <repository_url>


## mearge conflict : when two branches have made changes to the same line of code
## to resolve merge conflict
git status
git mearge --abort
git mearge --continue


## git rebase : to apply changes from one branch to another

git checkout feature
git rebase develop  


## git stash : to save changes temporarily
git stash       

git stash apply

git stash pop




```
