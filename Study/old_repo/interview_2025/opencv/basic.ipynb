## rotation

```python
import cv2
img = cv2.imread("elepent.jpg")
# cv2.imshow("original image", img)

width = img.shape[0]
hight = img.shape[1]
center = (hight//2, width//2)

metrix = cv2.getRotationMatrix2D(center, 50, 1)

rotate_img = cv2.warpAffine(img, metrix, (hight, width))

print(width, hight, center)
cv2.imshow("rotated image", rotate_img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

## shift and rotate

```python
import cv2
import numpy as np
img = cv2.imread("elepent.jpg")
# cv2.imshow("original image", img)

width = img.shape[0]
hight = img.shape[1]
center = (hight//2, width//2)
x = 50
y = 100
m = np.float32([[1,0,x], [0,1,y]])

# metrix = cv2.getRotationMatrix2D((50, 50), 50, 1)

rotate_img = cv2.warpAffine(img, m, (hight, width))

print(width, hight, center)
cv2.imshow("shifted image", rotate_img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

## intensity
* log transformation
* gamma transformation

```python
import cv2
import numpy as np
img = cv2.imread("elepent.jpg")
# cv2.imshow("gray image", img)
# log transformation 

def log_t(img):
    c = 255/np.log(1+np.max(img))
    # c is a scaling constant given by 255/log(1+m) , where m is the maximum pixel value in image
    lt = c*np.log(1+img)
    lt = np.array(lt, dtype=np.uint8)
    return lt



def gamma_t(img, gamma):
    gt = np.array(255*(img/255)**gamma,dtype=np.uint8)
    return gt

# new_img = log_t(img)
new_img = gamma_t(img, 0.2)
cv2.imshow("log tran", new_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## video 
* recorded video

```python
import numpy as np
import cv2

video_path = "video/v1.webm"
cap = cv2.VideoCapture(video_path)
while cap.isOpened():
    ret, frame =cap.read()
    if ret:
        img=cv2.resize(frame, (1024,720))
        cv2.imshow("video",img)
        if cv2.waitKey(25) & 0xff == ord('q'):
            break
    else:
        break
cap.release()
cv2.destroyAllWindows()
```

## time complexity 
* cv2.split() is a costly ops in terms of time. use numpy split

* split using cv2, use can see the time difference

```python
import cv2
import numpy as np
import time
img = cv2.imread("elepent.jpg")
img = cv2.resize(img, (1280,720))
start_time = time.time()
for i in range(100):
    r,g,b = cv2.split(img)

end_time=time.time()
print(end_time-start_time)
```
