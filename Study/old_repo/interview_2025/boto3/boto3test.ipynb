## to upload file

```python
import boto3
from botocore.exceptions import NoCredentialsError, ClientError
client = boto3.client("s3")
client.upload_file('local.txt',
                   Bucket='abc',
                   Key='upload/local_file.txt')

## Extra Argument

client.upload_file('image.png','abc','images/image.png',
                   ExtraArgs={
                       'ACL': 'public-read',
                       'ContentType' : 'image/png'
                   })
## using file object

with open('data.csv') as f:
    client.upload_fileobj(f, 'abc', 'data/data.csv')
print("upload via file object")



try:
    client.upload_file('local.txt',
                   Bucket='abc',
                   Key='upload/local_file.txt')
except FileNotFoundError:
    print("file not found")
except ClientError as e:
    print(e)
```

## for large files

```python
import boto3
from boto3.s3.transfer import TransferConfig

config = TransferConfig(
    multipart_threshold=1024 * 1024 * 100, # 100mb
    multipart_chunksize= 1024 * 1024 * 100, # 100 mb per part parallel uploads
    max_concurrency=5,
    use_threshold=True
)

s3 = boto3.cilent('s3')
s3.upload_file(Filename="large.zip",
               Bucket="abc",
               Key="uploads/large.zip",
               Config=config)
print("5gb file upload succesfully")
```

## streaming upload

```python
with open("largefile.zip", "rb") as f:
    s3.upload_fileobj(f, 'abc', 'uploads/largefile.zip',
                      Config=config)
    
```

## media files

```python
from boto3.s3.transfer import TransferConfig
import boto3

config = TransferConfig(
    multipart_threshold=1024*1024*100,
    multipart_chunksize=1024 * 1024 * 100,
    max_concurrency=8,
    use_threshold=True
)
s3 = boto3.client('s3')
s3.upload_file("movi.mp4",
               "abc",
               "media/video/movi.mp4"
               ExtraArgs={"ContentType":"video/mp4"},
               Config=config)
```

```python
Small media → upload_file

Large media → multipart upload_file

API upload → upload_fileobj

Streaming → Presigned URLs

Production → CloudFront
```
