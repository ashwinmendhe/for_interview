## mearging and join

```python
import pandas as pd

customer = {'customerid':[1,2,3],
            'name':['ramesh', 'suresh', 'manish']}
order = {'customerid':[1,2,4],
         'orderamt':[210, 230, 250]}
df_cust = pd.DataFrame(customer)
df_order = pd.DataFrame(order)
# print(df_cust)
# print(df_order)
## inner join
df_inner = pd.merge(df_cust, df_order, on='customerid', how="inner")

## outer join
df_outer = pd.merge(df_cust, df_order, on="customerid", how="outer")
df_outer

## left join
df_left = pd.merge(df_cust, df_order, on="customerid", how="left")
df_left
## right join
df_right = pd.merge(df_cust, df_order, on="customerid", how="right")
df_right

## cross
# df_cross = pd.merge(df_cust, df_order, on="custmerid", how="cross")
# df_cross
```

## concatinate
* vertically (row-wise)
* horizontally (colomn-wise)

```python
import pandas as pd

customer1 = {'customerid':[1,2,3],
            'name':['ramesh', 'suresh', 'manish']}
customer2 = {'customerid':[4,5,6],
            'name':['mangesh', 'sachin', 'ankit']}
df_cust1 = pd.DataFrame(customer1)
df_cust2 = pd.DataFrame(customer2)

## concatinate vertically means row wise here axis is 0

df_concat=pd.concat([df_cust1, df_cust2], ignore_index=True)
df_concat

## concatinate horizontally means coloum wise here axis is 1
df_concat_hori =pd.concat([df_cust1,df_cust2], axis=1, ignore_index=True)
df_concat_hori
```

## practicing large file with the help of genrator or pandas

```python
import pandas as pd

def read_line(filname):
    with open(filname, 'r') as file:
        for line in file:
            yield line
for line in read_line("now.csv"):
    print(line)
```

```python
## using genratoe for fized size chunk

def read_chunk(filename, chunk_size=1024*1024):
    with open(filename, 'rb') as f:
        while True:
            chunk = f.read(chunk_size)
            if not chunk:
                break
            yield chunk

for chunk in read_chunk("employee.csv"):
    print(chunk)
```

```python
## genrator using csv module

import csv
def csv_reader(filename):
    with open(filename, newline="") as f:
        reader = csv.reader(f)
        for row in reader:
            yield row
for row in csv_reader("employee.csv"):
    print(row)
```
