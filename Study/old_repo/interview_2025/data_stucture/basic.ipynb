Big O Notation
O(n) if number increased order of complaxity and nuber also increased

```python
## 0(n)

def get_squered_numbers(numbers):
    result = []
    for number in numbers:
        result.append(number * number)
    return result
# Big O Notation

numbers = [1, 2, 3, 4, 5]
print(get_squered_numbers(numbers))
# -*- coding: utf-8 -*- 
    
```

## Serch number
here complexity is 0(n)   iteration is len(n)

```python
def serch_number(number):
    for i in range(len(number)):
        if number[i] == 68:
            print(i)


serch_number([23,45,67,89,90,68,45,23,12])
```
