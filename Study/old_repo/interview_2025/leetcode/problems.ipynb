## sum of three no. is 0 , find thouse three numbers


```python

def sort_arr(arr):
    for i in range(len(arr)-1,0,-1):
        for j in range(i):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr

arr = [-1, 0, 1,2,-1,-4]
sorted_arr = sort_arr(arr)

def find_threeplet(sorted_arr):
    length = len(sorted_arr)

    res = []

    for i in range(length-2):

        if i > 0 and sorted_arr[i]==sorted_arr[i+1]:
            continue
        left = i+1
        right = length -1
        while left < right:
            total = sorted_arr[i] + sorted_arr[left] + sorted_arr[right]
            if total < 0:
                left +=1
            elif total > 0:
                right -=1
            else:
                res.append([sorted_arr[i],sorted_arr[left], sorted_arr[right]])
                while sorted_arr[left] == sorted_arr[left+1]:
                    left +=1
                while sorted_arr[right] == sorted_arr[right+1]:
                    right -= 1
                left +=1
                right -=1
    return res

print(sorted_arr)
print(find_threeplet(sorted_arr))
```

## 1. merge string 1 and string 2


```python
str1 = "abcd"
str2= "efghaaaaa"

def merge_str(str1,str2):
    result = ""
    i = 0
    while(i<len(str1) or i <len(str2)):
        if i < len(str1):
            result += str1[i]
        if i < len(str2):
            result += str2[i]
        i +=1
    return result
print(merge_str(str1, str2))
```

## 2. Sum of three no.

```python

def sum_three(arr, target):
    arr.sort()
    size = len(arr)
    
    for i in range(size-2):
        l = i+1
        r = size-1
        while l<r:
            if target == arr[i] + arr[l] + arr[r]:
                return arr[i] , arr[l] , arr[r]
            elif target < arr[i] + arr[l] + arr[r]:
                r -=1
            elif target > arr[i] + arr[l] + arr[r]:
                l += 1
            else:
                return False


arr = [1,4,6,8,10,13,45]
target = 22
print(sum_three(arr, target))
```

## 3. first and last position of number in array

```python
def position(arr, target):
    size = len(arr)
    first = -1
    last = -1
    for i in range(size):
        if target != arr[i]:
            continue
        if first == -1:
            first = i
        last = i
    return first, last

    


arr = [5,7,7,8,8,10]
# arr = [5,6,7,8,9,10]
target = 10
print(position(arr,target))
```

## longest substring

```python
def longest_substring(s):
    st = 0
    max_len = 0
    start_idx = 0
    d = {}

    for i, v in enumerate(s):
        # If character already seen in current window
        if v in d and d[v] >= st:
            st = d[v] + 1

        d[v] = i

        # Check if we found a longer substring
        if i - st + 1 > max_len:
            max_len = i - st + 1
            start_idx = st

    longest = s[start_idx : start_idx + max_len]
    return max_len, longest

print(longest_substring("abbcdabcbb"))
```

```python
def long(s):
    st = 0
    max_len = 0
    start_idx = 0
    d = {}
    for i,v in enumerate(s):
        if v in d and st <=d[v]:
            st = d[v] +1
        d[v] = i

        if i - st + 1 > max_len:
            max_len = i - st +1
            start_idx = st
    longest = s[start_idx:max_len+start_idx]

    return longest, max_len

print(long("abcdabcbb"))
```
