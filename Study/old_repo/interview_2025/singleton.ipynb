## Singleton Design pattern is : A Singleton in Python is a design pattern where only one instance of a class can be created. No matter how many times you create the object, you will always get the same instance.

```python
## eg : This below getting two instane of same class to overcome this we need singleton pattern
## here we have created two instance of same class , and when we match memory objects are
## different for both , and that is not exceptable
class ControllTower:
    def __init__(self):
        print("initialize Tower")

tower1 = ControllTower()
tower2 = ControllTower()
print(tower1 is tower2)
```

```python
## When we invoke , we dont want to create instance every time
## __init__ constructor : Controls the Initiazation of an instance
## so we create __new__ before the init : new Controls how we create an instance
class ControllTower:
    _instance = None
    def __new__(cls):
        if cls._instance is None: ## check if instance already created
            cls._instance = super().__new__(cls)
            print("initialize Tower")
        return cls._instance

tower1 = ControllTower()
tower2 = ControllTower()
print(tower1 is tower2)
```
