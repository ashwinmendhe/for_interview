```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def get_avg(self):
        sum = 0
        for val in self.marks:
            sum += val
        print(f"{self.name} avarage is {sum//len(self.marks)}")

    @staticmethod
    def hello():
        print("hello")

s1 = Student("ashwin", [99,85, 86])
s1.get_avg()
s1.hello()



```

## super method

```python
class Car:
    def __init__(self, type):
        self.type = type
    
    @staticmethod
    def start():
        print("start")
    
    @staticmethod
    def stop():
        print("stop")

class ToyotaCar(Car):
    def __init__(self, name, type):
        self.name = name
        super().__init__(type)
        super().start()
        super().stop()

car1 = ToyotaCar("abc", "electric")
# car1.start()
# car1.stop()
car1.name
car1.type

```

* init : initialize the instance (set up values)
* new : creates a new instance (like constucture in other program)

```python
## init

class Person:
    def __init__(self, name, age):
        print("init function")
        self.name = name
        self.age = age

p1 = Person("yesh", 25)
print(p1.name, p1.age)
```

```python
## new: here fist need to crate class instance , which is making here using new

class Person:

    def __new__(cls, *args, **kwarg):
        print("insde new")
        instance = super().__new__(cls)
        print("Instance created", instance)
        return instance

    def __init__(self, name, age):
        print("init function")
        self.name = name
        self.age = age

p1 = Person("yesh", 25)
print(p1.name, p1.age)
```

## Singleton Method 1

```python
import random
class Game:
    _instance = None
    def __init__(self):
        words = ["cat", "bat", "dog", "apple", "orange", "mat"]
        self.result = random.choice(words)
        Game._instance = self
    def guess(self, result):
        if result == self.result:
            print("its match")
        else:
            print("no match")
        print(f"sys word is {self.result} and your word {result}")

    @staticmethod
    def getInstance():
        if Game._instance == None:
            Game()
        return Game._instance


p1 = Game.getInstance()
p2 = Game.getInstance()
p1.guess("dog")
p2.guess("cat")
print(p1 is p2)

```
