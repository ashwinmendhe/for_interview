## 2 What is decorator 
A decorator in Python is a function that modifies the behavior of another function or class — without permanently changing its code.

eg. 

```python
def decorator(func):
    def wrapper():
        print("wrapper executed this before {}".format(func.__name__))
        return func()
    print("decorator executed this when function is defined")
    return wrapper

def show():
    print("show function executed")
decorator_show = decorator(show)
decorator_show()

```

```python
def deco(fu1):
    def wrap():
        print("wrap executed this before {}".format(fu1.__name__))
        return fu1()
    print("deco executed this when function is defined")
    return wrap

@deco
def disp():
    print("disp function executed") 

disp()
```

## 4 Iterator and genrator

genrators are iterators which can execute only once

interator is an object which contains a countable nuber of values and it is used iterate over iterable objects like list tuples , sets etc

```python
def sqr(n):
    for i in range(n):
        yield i*i 
a = sqr(3)
print(next(a))
print(next(a))
print(next(a))
```

```python
iter_list = iter([1,2,3,4,5])
print(next(iter_list))
print(next(iter_list))
# print(next(iter_list))
```

## 6. self

## recurssing and revers 

```python
def reverse_list(lst):
    if len(lst) == 0:
        return []
    else:
        return [lst[-1]] + reverse_list(lst[:-1])
    
print(reverse_list([1, 2, 3, 4, 5]))
```

## Draw pattern 

```python
n = int(input("Enter number of rows: "))
for i in range(0,n):
    for j in range(0, i+1):
        print("*", end="")
    print()
```
