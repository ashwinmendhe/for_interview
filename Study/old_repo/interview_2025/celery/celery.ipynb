## Celery : Distributed Task queue
* Asynchronous task queue

* RabbitMQ is the task queue is similar like redis, it works seamlessly, by creating instance of rabbitmq using docker (runs on 15672)
* celery takes thing from your task q(rabbit mq) and execute the same.

## celery program

* celery server
* lets say file name : main.py

```python
from celery import Celery
import time

app = Celery('cel_main', broker = 'pyamqp://')
## pyamqp: python advanced message queuing protocol, by defaults its uses rabbitmq
@app.task
def TaskQueue(message):
    time.sleep(5)
    print(f"taskqueue: {message}")

@app.task
def write_log(logs):
    print(f"writing logs: {logs}")


```

#### To run server
* celery -A main worker --loglevel=INFO

* celery -A main worker --concurrency=2 --loglevel=INFO ## When we mentioed concurrency its create 2 celery instance and run two task parallely, we can increase this to 20 to 40

#### When celery runs after ther our client program can call by celery and executed

* celery client

```python
from main import TaskQueue
TaskQueue.delay("hi") ## delay is for to saved in rabbitmq
write_log.delay("logs")
```

✅ 1. Celery Beat

Celery Beat is a scheduler.

What it does:

Runs tasks at fixed intervals (like cron)

Sends scheduled tasks to the Celery queue

Your Celery workers pick them and execute

When to use Celery Beat:

Run a task every 10 seconds

Send daily reports

Cleanup database every night

Fetch API data every hour

✅ 2. Celery Worker

Celery Workers are the actual executors of tasks.

What they do:

Receive tasks from queue (Redis/RabbitMQ)

Execute the task logic

Can run many tasks in parallel (multiple workers)

When to use Workers:

Asynchronous tasks (send email, process file)

Heavy background operations

Real-time task execution

Command:
celery -A app worker --loglevel=info
