## Advance sql query

window , dense, rank

```python
select s.* from student as s;

## its create one more coloum with same value as max_marks
## max(marks) is the aggregate function, over() is the clause, sql treate max as not the aggregate 
## function its a a window function , so its create window of records, in over no coumn mentions so
## its treated as one window for all record
select s.* , max(marks) over() as max_marks from student as s;


## max marks for each city, i have to use here partition by(caluse) {coloumn name}, its gives max_marks for 
## city and group as window. 
select s.* , max(marks) over(partition by city) as max_marks from student as s;



```

* row_number(), rank, dense, lag , lead

```python
## its creating rn as new column as gives row_number for all rows in coulomn, its create one window
## for all records 
select s.*, row_number() over() as rn from student as s;

## its for grouping row number, and create multiple window for records

select s.* , row_number() over(partition by city) as rn from student as s;


```

## Que : fetch the first two employee of each department
* using window rank row_number function

```python
## over is the window query and inside that its grouped by department and order by emp_id
### this below giving row_numbers of each department with sorted emp_id

select e.* , row_number() over(partition by department order by emp_id) as rn from employees as e;

## for 1st two record

select * from (select e.* , row_number() over(partition by department order by emp_id) as rn from employees as e;
) as x where x.rn < 3;
```

## Rank

```python
## its giving rank of record according to salary 
select e.* , rank() over(partition by department order by salary desc) as rnk from employee as e;
```

## Que: top 3 employee salary 

```python
select * from
(select e.* , rank() over(partition by department order by salary desc) as rnk from employee as e)
as x where x.rnk <3;
```

## dense_rank


```python
## in rank if duplicate rows are there with same salary its skip next rank and increase , 
## but in dense_rank its not skip any rank if there is same salary record exist.

select e.*, rank() over(partition by department order by salary desc) as rnk,
dense_rank() over(partition by department order by salary desc) as DENSE_rnk from employee as e; 
```

## combination of rank, dense_rank, row_number


```python
select * from 
(select e.*, rank() over(partition by department order by salary desc) as rnk , 
dense_rank() over(partition by department order by salary desc) as DENS_RANK from employee as e) as x 
where x.rnk < 4;
```

## Lag

```python
## in lag its check prvious employee salary
select e.*, lag(salary) over(partition by department order by emp_id desc) as prev_salary from employee as e;

## lag(salary, 2, 0)  : 2 means its checked 2 steps previous, 0 is by default if value not present its give 0
select e.*, lag(salary, 2, 0) over(partition by department order by emp_id desc) as prev_salary from employee as e;
```

## Lead

```python
select e.* , lead(salary) over(partition by department order by emp_id desc) as next_salary from employee as e;

select e.* , lead(salary, 2, 0) over(partition by department order by emp_id desc) as next_salary from employee as e;
```

```python
select e.* , lag(e.salary) over(partition by e.department order by e.emp_id desc) as pre_salary,
case when e.salary > lag(e.salary) over(partition by e.department order by e.emp_id desc) then "higher that pre salary"
	when e.salary < lag(e.salary) over(partition by e.department order by e.emp_id desc) then "lower than previous salary"
    when e.salary = lag(e.salary) over(partition by e.department order by e.emp_id desc) then "equal than previous salary"
	end sal_range
from employee as e;
```
