```python
Browser A / B: Peers sending audio/video streams.

Signaling Server: Uses WebSocket to exchange connection setup messages (SDP + ICE).

STUN/TURN Servers: Handle NAT/firewall traversal.

PeerConnection: Direct encrypted media/data link.
```

```python
Both browsers connect to a WebSocket signaling server.

Browser A calls getUserMedia() to capture audio/video.

A creates an SDP offer and sends it to B through the signaling server.

B receives it, creates an SDP answer, and sends it back.

Both exchange ICE candidates (network info).

Once ICE connection succeeds → direct P2P connection established.

Audio/video streams flow directly between browsers (no server in between).
```

## SDP

```python
“SDP in WebRTC contains everything required to negotiate a peer-to-peer connection:
supported codecs, encryption keys, ICE candidates (network paths), transport protocols, 
media directions, bandwidth constraints, and RTP parameters.
Both peers exchange SDP via signaling (like WebSocket/REST), 
and once negotiation is done the media flows directly, bypassing the signaling server.”
```

## Architechture

```python
Camera → GStreamer Pipeline → kvssink → Kinesis Video Stream
                         ↑
            FastAPI controls the pipeline (start/stop/status)
```

“To stream video using AWS Kinesis Video Streams, I normally use AWS KVS Producer SDK or the GStreamer kvssink plugin. The pipeline captures the camera feed, encodes it to H.264 using x264enc, and pushes frames directly to a KVS stream via kvssink. On top of that, I build a FastAPI service that manages the pipelines: starting/stopping streams, generating stream names, and monitoring health. FastAPI itself doesn't serve the video; it just orchestrates the streaming, whereas the actual media transport happens inside GStreamer. This architecture is scalable, low-latency, and AWS-optimized.”
