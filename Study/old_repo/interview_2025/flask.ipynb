1️⃣ What is Flask?

Flask is a lightweight WSGI web framework. It’s simple, flexible, and modular, used for small to medium applications.


```python
from flask import Flask
app = Flask(__name__)

@app.route('/')
def home():
    return 'Hello Flask'

if __name__ == '__main__':
    app.run(debug=True)
```

2️⃣ How do you handle routes and dynamic URLs?
Use @app.route decorator


```python
@app.route('/user/<name>')
def user(name):
    return f'Hello {name}'
```

3️⃣ Explain Blueprints
Blueprints organize routes into modules.


```python
from flask import Blueprint
user_bp = Blueprint('user', __name__)

@user_bp.route('/profile')
def profile():
    return 'Profile Page'

app.register_blueprint(user_bp)
```
