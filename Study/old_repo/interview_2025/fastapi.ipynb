1️⃣ Why is FastAPI fast?
It uses Starlette (ASGI) and Pydantic for async and type validation.


* Client → ASGI Server → Middleware (before)
        → Routing → Dependencies → Validation
        → Path Operation Function
        → Response Model Serialization
        → Middleware (after) → ASGI → Client


```python
Interview-Style Summary

The FastAPI request-response lifecycle:

Request received by ASGI server

Middleware pre-processing

Route matching

Dependency injection

Data validation (Pydantic)

Execute path operation function

Response serialization

Middleware post-processing

Response returned to client
```

2️⃣ Example route


```python
from fastapi import FastAPI
app = FastAPI()

@app.get('/')
def root():
    return {'message': 'Hello FastAPI'}
```

3️⃣ Request body and validation


```python
from pydantic import BaseModel
class Item(BaseModel):
    name: str
    price: float

@app.post('/items')
def create_item(item: Item):
    return item
```

4️⃣ Dependency Injection


```python
Use Depends.

Example:
from fastapi import Depends

def get_db():
    return 'DB Connected'

@app.get('/data')
def data(db=Depends(get_db)):
    return db
```

5️⃣ Background tasks


```python
from fastapi import BackgroundTasks

def notify(msg):
    print(msg)

@app.post('/send')
def send(background_tasks: BackgroundTasks):
    background_tasks.add_task(notify, 'Email sent')
```
