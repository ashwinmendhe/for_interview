```python
def merge_overlaps(intervals):
    # Step 1: Sort intervals by start time
    intervals.sort(key=lambda x: x[0])

    merged = [intervals[0]]

    # Step 2: Loop through and merge if overlapping
    for current in intervals[1:]:
        prev_start, prev_end = merged[-1]
        curr_start, curr_end = current

        if curr_start <= prev_end:  # Overlapping
            merged[-1] = (prev_start, max(prev_end, curr_end))  # Merge
        else:
            merged.append(current)

    return merged


# Example
times = [(1, 5), (4, 7), (8, 10), (9, 12)]
result = merge_overlaps(times)
print(result)
```

## remove string in paranthesis

```python
def parse_string(s: str):
    stack = 0

    # Check if parentheses are balanced
    for ch in s:
        if ch == '(':
            stack += 1
        elif ch == ')':
            stack -= 1
            if stack < 0:  # more closing than opening
                return None

    if stack != 0:  # unbalanced parentheses
        return None

    # If valid, remove all parentheses
    result = ""
    for ch in s:
        if ch != '(' and ch != ')':
            result += ch

    return result


# Example usage
inp = ["abc(rtyh(dfgh)gghj(sdfgh)iuytre)", "adfgff)dfgh(dfghj)(dfghj"]

out = []
for s in inp:
    out.append(parse_string(s))

print(out)
```

```python
def binary_rotated(arr, target):
    low = 0 
    high = len(arr) -1
    while low<=high:
        mid = (low+high)//2
        if arr[mid] == target:
            return mid, arr[mid]
        if arr[low] <= arr[mid]:
            if arr[low] <= target < arr[mid]:
                high = mid -1
            else:
                low = mid +1
        else:
            if arr[mid] < target <= arr[high]:
                low = mid+1
            else:
                high = mid -1
    return -1
         

        




arr = [40,50,60,70,10,20,30]
target = 50
print(binary_rotated(arr, target))
```

## Min meeting rooms

```python
def min_meeting_rooms(intervals):
    if not intervals:
        return 0

    # Separate start and end times
    starts = sorted([i[0] for i in intervals])
    ends = sorted([i[1] for i in intervals])

    s, e = 0, 0
    rooms = 0
    max_rooms = 0

    while s < len(intervals):
        if starts[s] < ends[e]:
            rooms += 1
            s += 1
        else:
            rooms -= 1
            e += 1
        max_rooms = max(max_rooms, rooms)

    return max_rooms

# Example
intervals = [[1,3], [1,3], [1,5]]
print(min_meeting_rooms(intervals))  
```

```python
def flat(arr):
    result = []
    for num in arr:
        if isinstance(num, list):
            result.extend(flat(num))
        else:
            result.append(num)
    return result

arr = [20, 30,[40,50, [1,3,0]], [45,5,67]]
print(sorted(flat(arr), reverse=True))  
```

```python
def sort_array(arr):
    for i in range(len(arr)-1, 0, -1):
        for j in range(i):
            if arr[j] > arr[j+1]:
                arr[j+1], arr[j] = arr[j], arr[j+1]

    return arr

arr = [5,3,8,6,7,2]

print(sort_array(arr))

    
```

```python
def sort_code(s: str) -> str:
    freq = [False] * 26  # track which characters appear

    # Mark characters as present
    for ch in s:
        freq[ord(ch) - ord('a')] = True

    # Build sorted output
    result = []
    for i in range(26):
        if freq[i]:
            result.append(chr(i + ord('a')))

    return ''.join(result)
```

```python



def longest_common_prefix(strs):
    if not strs:
        return ""

    # Start with the first string
    prefix = strs[0]

    for s in strs[1:]:
        # Reduce prefix until it matches the start of s
        while not s.startswith(prefix):
            prefix = prefix[:-1]
            if not prefix:
                return ""
    return prefix

a = ["flower", "flown", "flow", "floight"]
b = ["car", "train", "bike"]
print(longest_common_prefix(a))
```

```python
def longest_common_string(strs):
    # Take the shortest string as reference
    shortest = min(strs, key=len)
    n = len(shortest)

    best = ""

    # Check all substrings of the shortest string (largest to smallest)
    for i in range(n):
        for j in range(i+1, n+1):
            sub = shortest[i:j]

            # Check how many strings contain this substring/prefix
            count = sum(sub in s for s in strs)

            # If at least 2 or more strings share it
            if count >= 2 and len(sub) > len(best):
                best = sub

    return best

a = ["aflower", "flown", "flow", "floight"]
print(longest_common_string(a))
```

```python
def long(s):
    st = 0
    max_len = 0
    start_idx = 0
    d = {}
    for i,v in enumerate(s):
        if v in d and st <=d[v]:
            st = d[v] +1
        d[v] = i

        if i - st + 1 > max_len:
            max_len = i - st +1
            start_idx = st
    longest = s[start_idx:max_len+start_idx]

    return longest, max_len

print(long("abcdabcbb"))
```

## Large file 5gb file using genrator

```python

## using generator for small file read line by line (only one line stays in memory)
def read_large_file(filename):
    with open(filename, 'r') as f:
        for line in f:
            yield line

for line in read_large_file("data.csv"):
    print(line)
```

```python
## genrator using fixed size chunks

def read_in_chunck(filename, chunk_size = 1024*1024):
    with open(filename, 'rb') as f:
        while True:
            chunk = f.read(chunk_size)
            if not chunk:
                break
            yield chunk

for chunk in read_in_chunck("data.bin"):
    print(chunk)
```

```python
# genrator for csv
import csv
def csv_genrator(filename):
    with open(filename, newline="") as f:
        reader = csv.reader(f)
        for row in reader:
            yield row

for row in csv_genrator("data.csv"):
    print(row)
```

```python
def InsertionSort(A):
    for j in range(1, len(A)):
        key = A[j]
        i = j-1
    while (i >= 0) and (A[i] > key):
        A[i+1] = A[i]
        i = i-1
    A[i+1] = key

input = [8,3,9,15, 29, 7, 10]
InsertionSort(input)
print(input)
```
