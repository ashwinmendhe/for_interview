## Filter

* by using filter we can fiter even no.

```python
nums = [3,2,6,8,2,9,5,4]
def is_even(n):
    return n%2 ==0

evens = list(filter(is_even,nums))
evens
```

* by using lamda

```python
nums = [3,2,6,8,2,9,5,4]
evens = list(filter(lambda n: n%2==0 ,nums))
evens
```

## map

```python
nums = [3,2,6,8,2,9,5,4]
def update(n):
    return n+2

evens = list(filter(lambda n: n%2==0 ,nums))
Doubles = list(map(update,evens))
Doubles
```

* using lambda

```python
nums = [3,2,6,8,2,9,5,4]

evens = list(filter(lambda n: n%2==0 ,nums))
Doubles = list(map(lambda n : n+2,evens))
print(evens)
print(Doubles)
```

## reduce

```python
from functools import reduce

nums = [3,2,6,8,2,9,5,4]
def add_all(a,b):
    return a+b

evens = list(filter(lambda n: n%2==0 ,nums))
Doubles = list(map(lambda n : n+2,evens))
print(evens)
print(Doubles)
sum_n = reduce(add_all,Doubles)
print(sum_n)
```

* lambda for reduce

```python
from functools import reduce

nums = [3,2,6,8,2,9,5,4]

evens = list(filter(lambda n: n%2==0 ,nums))
Doubles = list(map(lambda n : n+2,evens))
print(evens)
print(Doubles)
sum_n = reduce(lambda a,b : a+b ,Doubles)
print(sum_n)
```
