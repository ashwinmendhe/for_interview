## shallo or deep copy

```python
import copy
a = [[1,2],[3,4]]
s= copy.copy(a)
d = copy.deepcopy(a)
a[0][0] = 90
print(s)
print(d)
```

## iterator or genrator

```python
def gen():
    for i in range(3):
        yield i
for n in gen():
    print(n)
```

3️⃣ Decorators
Decorators modify function behavior without changing code.


```python
def log_decorator(func):
    def wrapper():
        print(f'Calling {func.__name__}')
        func()
    return wrapper

@log_decorator
def greet():
    print('Hello!')

greet()
```

4️⃣ Global Interpreter Lock (GIL)
GIL allows only one thread to execute Python bytecode at a time. Use multiprocessing for CPU-bound tasks.


5️⃣ @classmethod, @staticmethod, and instance methods
Instance → self; Classmethod → cls; Staticmethod → no self/cls.


```python
class Example:
    def inst(self): return 'instance'
    @classmethod
    def cls(cls): return 'class'
    @staticmethod
    def stat(): return 'static'
```

6️⃣ Mutable vs Immutable
Mutable (list, dict, set) can change; Immutable (str, tuple, int) cannot.


7️⃣ Context Managers
Use 'with' to manage resources automatically.


```python
# Example:
with open('file.txt', 'w') as f:
    f.write('Hello')
```

8️⃣ Dataclasses
Simplify class creation.


```python
from dataclasses import dataclass
@dataclass
class User:
    name: str
    age: int
```
