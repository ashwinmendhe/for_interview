Arithmatic error


* ZeroDivisionError	Division by zero
* OverflowError	Result too large to represent
* FloatingPointError	Floating-point calculation error

Lookup Errors

* IndexError	Invalid list or tuple index
* KeyError	Accessing a non-existing dictionary key

🔹 File and I/O Errors
* FileNotFoundError	File not found
* IOError	Input/output operation failure
* PermissionError	Lack of file access permission

🔹 Import and Module Errors
* ImportError	Import fails
* ModuleNotFoundError	Module doesn’t exist

🔹 Runtime and System Errors
* RuntimeError	General runtime issue
* RecursionError	Recursion depth exceeded
* MemoryError	Out of memory

```python
try:
    # Trigger different errors for testing
    num = int(input("Enter a number: "))     # Might raise ValueError
    result = 10 / num                        # Might raise ZeroDivisionError
    data = {"name": "Alice"}
    print(data["age"])                       # Might raise KeyError

except ValueError:
    print("❌ Invalid input! Please enter a number.")
except ZeroDivisionError:
    print("❌ You cannot divide by zero.")
except KeyError as e:
    print(f"❌ Missing key in dictionary: {e}")
except Exception as e:
    print(f"⚠️ Some other error occurred: {e}")
finally:
    print("✅ Execution complete — cleaning up resources if needed.")
```
