### Problem 3
Given a list of intervals, merge overlapping intervals.

```python
def merge_intervals(intervals):
    if not intervals: return []
    intervals.sort(key=lambda x: x[0])
    merged = [intervals[0]]
    for s,e in intervals[1:]:
        last_s, last_e = merged[-1]
        if s <= last_e:
            merged[-1] = (last_s, max(last_e, e))
        else:
            merged.append((s,e))
    return merged

print(merge_intervals([(3,5), (2,6), (9,10), (11,12)] ))
```

### Problem 1
Create a timer decorator that prints execution time.

```python
import time
from functools import wraps

def timer(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        s=time.time(); r=func(*args, **kwargs); print(f'Elapsed: {time.time()-s:.4f}s'); return r
    return wrapper

@timer
def work(n):
    sum(i*i for i in range(n))

work(100000)
```

### Problem 2
Create a decorator that caches results (simple memoize).

```python
def memoize(func):
    cache = {}
    def wrapper(*args):
        if args in cache:
            return cache[args]
        cache[args] = func(*args)
        return cache[args]
    return wrapper

@memoize
def fib(n):
    if n<2: return n
    return fib(n-1)+fib(n-2)

print(fib(30))
```

### Problem 3
Write an access-control decorator that checks a user role in kwargs.

```python
from functools import wraps

def require_role(role):
    def deco(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            if kwargs.get('role') != role:
                raise PermissionError('Forbidden')
            return func(*args, **kwargs)
        return wrapper
    return deco

@require_role('admin')
def delete_all(*args, **kwargs):
    return 'deleted'

try:
    print(delete_all(role='user'))
except Exception as e:
    print(e)
print(delete_all(role='admin'))
```

### Problem 1
Implement LRU cache using OrderedDict.

```python
from collections import OrderedDict
class LRU:
    def __init__(self, cap):
        self.cap=cap; self.od=OrderedDict()
    def get(self,k):
        if k not in self.od: return -1
        self.od.move_to_end(k)
        return self.od[k]
    def put(self,k,v):
        self.od[k]=v; self.od.move_to_end(k)
        if len(self.od)>self.cap: self.od.popitem(last=False)

c=LRU(2)
c.put(1,1);c.put(2,2);print(c.get(1));c.put(3,3);print(c.od)
```

### Problem 3
Add TTL (time-to-live) to a simple cache (dict + timestamps).

```python
import time
class TTLCache:
    def __init__(self, ttl):
        self.ttl=ttl; self.store={}
    def put(self,k,v):
        self.store[k]=(v,time.time())
    def get(self,k):
        val=self.store.get(k)
        if not val: return None
        v,t=val
        if time.time()-t>self.ttl:
            del self.store[k]; return None
        return v

c=TTLCache(0.1); c.put('a',1); import time; time.sleep(0.2); print(c.get('a'))
```

### Problem 2
Implement retry logic decorator for transient failures.

```python
import time
from functools import wraps

def retry(times=3, delay=0.1):
    def deco(f):
        @wraps(f)
        def w(*a, **k):
            for i in range(times-1):
                try: return f(*a, **k)
                except Exception:
                    time.sleep(delay)
            return f(*a, **k)
        return w
    return deco

@retry(3)
def flaky(x):
    import random
    if random.random()<0.8: raise Exception('fail')
    return 'ok'

try: print(flaky(1))
except: print('still failed')
```

Binary search to find element in sorted list.

```python
def bs(a,x):
    l,r=0,len(a)-1
    while l<=r:
        m=(l+r)//2
        if a[m]==x: return m
        if a[m]<x: l=m+1
        else: r=m-1
    return -1

print(bs([1,3,5,7],5))
```

### Problem 1
Longest substring without repeating characters (sliding window).

```python
def longest_unique(s):
    seen={}
    start=0; maxlen=0
    for i,ch in enumerate(s):
        if ch in seen and seen[ch]>=start:
            start=seen[ch]+1
        seen[ch]=i
        maxlen=max(maxlen, i-start+1)
    return maxlen

print(longest_unique('abcabcbb'))
```

### Problem 2
Longest palindromic substring (expand center).

```python
def longest_pal(s):
    def expand(l,r):
        while l>=0 and r<len(s) and s[l]==s[r]: l-=1; r+=1
        return s[l+1:r]
    res=''
    for i in range(len(s)):
        for part in (expand(i,i), expand(i,i+1)):
            if len(part)>len(res): res=part
    return res

print(longest_pal('babad'))
```

### Problem 1
Rotate an NxN matrix 90 degrees clockwise in-place.

```python
def rotate(mat):
    n=len(mat)
    for i in range(n//2):
        for j in range(i, n-1-i):
            mat[i][j], mat[n-1-j][i], mat[n-1-i][n-1-j], mat[j][n-1-i] = \
            mat[n-1-j][i], mat[n-1-i][n-1-j], mat[j][n-1-i], mat[i][j]

m=[[1,2,3],[4,5,6],[7,8,9]]
rotate(m)
print(m)
```

```python
def analyze(log):
    from collections import Counter
    cnt=Counter()
    for line in log.splitlines():
        if 'ERROR' in line: cnt['ERROR']+=1
        elif 'WARNING' in line: cnt['WARNING']+=1
        elif 'INFO' in line: cnt['INFO']+=1
    return cnt

log='INFO a\nERROR b\nWARNING c\nERROR d'
print(analyze(log))
```

Problem A (Decorator + OOP): create @timed method decorator inside a class and use it.

```python
import time
from functools import wraps
class C:
    def timed(f):
        @wraps(f)
        def w(*a,**k):
            s=time.time(); r=f(*a,**k); print('took',time.time()-s); return r
        return w
    @timed
    def work(self, n):
        sum(i*i for i in range(n))

C().work(100000)
```

### Problem 2
Problem B (DSA): given array, find longest subarray with sum = k (use hashmap).

```python
def longest_subarray_with_sum(arr,k):
    seen={0:-1}; s=0; ans=0
    for i,v in enumerate(arr):
        s+=v
        if s-k in seen:
            ans=max(ans, i-seen[s-k])
        if s not in seen:
            seen[s]=i
    return ans
print(longest_subarray_with_sum([1,2,3,-3,4],6))
```

### Problem 3
Problem C (Concurrency): implement producer-consumer using queue.Queue.

```python
from queue import Queue
from threading import Thread
import time

def producer(q):
    for i in range(5):
        q.put(i); time.sleep(0.01)
    q.put(None)

def consumer(q):
    while True:
        item=q.get()
        if item is None: break
        print('consumed', item)

q=Queue()
Thread(target=producer,args=(q,)).start()
Thread(target=consumer,args=(q,)).start()
time.sleep(0.2)
print('done')
```
