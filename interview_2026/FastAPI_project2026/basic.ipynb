```python
class Product:
    id: int
    name: str
    description: str
    price: float
    quantity: int

    def __init__(self, id: int, name: str, description: str, price: float, quantity: int):
        self.id = id
        self.name = name
        self.description = description
        self.price = price
        self.quantity = quantity

products = [
    Product(1, "phone", "budget phone", 99, 10 ),
    Product(2, "laptop", "this is laptop", 9999.123, 2 ),
]
def get_all_products(products):
    return products
print(get_all_products(products))
```
